# EDGX Burst Explorer
Interactive Plotly notebook for normalized bursts


```python
import pandas as pd
import plotly.express as px
from pathlib import Path
norm_path = Path('reports/diagnostics/edgx_bursts/normalized_points.parquet')
norm = pd.read_parquet(norm_path)
if 'burst_id' not in norm.columns:
    norm['burst_id'] = norm['date'].astype(str) + '_' + norm['start'].astype(str)
clusters = pd.read_csv('reports/diagnostics/edgx_bursts/burst_clusters.csv')
norm = norm.merge(clusters[['burst_id','cluster']], on='burst_id', how='left')
print(f'Loaded {len(norm)} points across {norm.burst_id.nunique()} bursts')
```

```python
sample = norm.sample(500000, random_state=0) if len(norm) > 500000 else norm
sample['cluster_str'] = sample['cluster'].astype(str)
fig = px.scatter(sample, x='norm_x', y='norm_y', color='cluster_str', hover_data=['burst_id','date'],
                 title='Normalized bursts (sample)')
fig.update_traces(marker=dict(size=2, opacity=0.35))
fig.update_layout(width=900, height=600)
fig.show()
```

```python
cluster_id = 0
sub = norm[norm['cluster'] == cluster_id]
sub_sample = sub.sample(300000, random_state=0) if len(sub) > 300000 else sub
sub_sample['date_str'] = sub_sample['date'].astype(str)
fig = px.scatter(sub_sample, x='norm_x', y='norm_y', color='date_str', hover_data=['burst_id'],
                 title=f'Cluster {cluster_id} detail')
fig.update_traces(marker=dict(size=2, opacity=0.35))
fig.update_layout(width=900, height=600)
fig.show()
```
